```python
import pandas as pd
```

```python
df = pd.read_csv("messy_sales_data.csv")  #read file
```

```python
print(df.isna().sum())
```

```python
print(df.duplicated().sum())                         # Step 6
```

```python
df = df.drop_duplicates()                            # Step 7
```

```python
df = df.dropna(subset=["customer_name", "order_date"]) 
```

```python
df["quantity"] = df["quantity"].fillna(df["quantity"].median())   # Step 9
```

```python
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].mean())
```

```python
df["rating"] = df["rating"].fillna(df["rating"].mean())
```

```python
df["payment_method"] = df["payment_method"].fillna("Unknown")
```

```python
df["city"] = df["city"].fillna("Unknown")
```

```python
df["email"] = df["email"].fillna("not_provided")
```

```python
print(df.isna().sum())   
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], format="mixed", dayfirst=True)
```

```python
df["city"] = df["city"].str.strip().str.title()  
```

```python
df.to_csv("cleaned_sales_data.csv", index=False) 
```
