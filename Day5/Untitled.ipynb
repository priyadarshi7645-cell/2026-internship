```python
import pandas as pd
```

```python
df = pd.read_csv("messy_orders.csv")  #read file
```

```python
print(df.isna().sum())    #missing values find 
```

```python
print(df.duplicated().sum())                         # duplicate identify
```

```python
df = df.drop_duplicates()      #duplicate remove 
```

```python
df["quantity"] = df["quantity"].fillna(df["quantity"].median())   
```

```python
df["rating"] = df["rating"].fillna(df["rating"])
```

```python
df["order_status"]=df["order_status"].str.title()
```

```python
df["unit_price"] = df["unit_price"].astype(str)

df["unit_price"] = df["unit_price"].str.replace(
    r'₹|Rs\.|,|\s', '', regex=True
)
```

```python
df["email"]=df["email"].str.title()
```

```python
df["age"] = df["age"].fillna(df["age"])
```

```python
df["ship_date"] = df["ship_date"].fillna(df["ship_date"])
```

```python
df["email"] = df["email"].fillna(df["email"])
```

```python
df["customer_name"]=df["customer_name"].str.title()
```

```python
# Quantity cleaning
df["quantity"] = pd.to_numeric(df["quantity"], errors="coerce")

# Unit price cleaning
df["unit_price"] = (
    df["unit_price"].astype("string")
    .str.replace(r'₹|Rs\.?|,|\s', '', regex=True)
)

df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")

# Fill missing values
df["quantity"] = df["quantity"].fillna(df["quantity"].median())
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].median())

# Create total column
df["total"] = (df["quantity"] * df["unit_price"]).round().astype("Int64")

# Display output
df[["quantity", "unit_price", "total"]]
```

```python
df["payment_method"]=df["payment_method"].str.title()
```

```python
df["rating"] = pd.to_numeric(df["rating"], errors="coerce").round().astype("Int64")
```

```python
df.to_csv("cleaned_sales_data.csv", index=False)
```
