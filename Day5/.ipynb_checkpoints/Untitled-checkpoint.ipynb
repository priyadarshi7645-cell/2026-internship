```python
import pandas as pd
```

```python
df = pd.read_csv("messy_orders.csv")  #read file
```

```python
print(df.head(3))  #show first 5 records
```

```python
print(df.shape)  #total rows and columns
```

```python
df.info() #columns details
```

```python
print(df.isna().sum())    #missing values find 
```

```python
print(df.duplicated().sum())                         # duplicate identify
```

```python
df = df.drop_duplicates()      #duplicate remove 
```

```python
df["quantity"] = df["quantity"].fillna(df["quantity"].median())   
```

```python
df["rating"] = df["rating"].fillna(df["rating"].mean())
```

```python
df["order_status"]=df["order_status"].str.title()
```

```python
df["unit_price"] = df["unit_price"].astype(str)

df["unit_price"] = df["unit_price"].str.replace(
    r'₹|Rs\.|,|\s', '', regex=True
)
```

```python
df["email"]=df["email"].str.title()
```

```python
df["age"] = df["age"].fillna(df["age"])
```

```python
df["ship_date"] = df["ship_date"].fillna(df["ship_date"])
```
